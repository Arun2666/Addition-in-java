```python
class Atm:

  def __init__(self):
    self.pin=" "
    self.balance=0
    self.menu()

  def menu(self):
    user_input= input("""
    hi how can i help you?
    1.press 1 to create pin
    2.press 2 to change pin
    3.press 3 to check balance
    4.press 4 to withdraw
    5.anytjing else to exit
    """)

    if user_input == "1":
      self.create_pin()
    elif user_input == "2":
      self.change_pin()
    elif user_input == "3":
      self.check_balance()
    elif user_input == "4":
      self.withdraw()
    else:
      print("Enter a vailt key" )

  def create_pin(self):
    user_pin = input("Enter pin")
    self.pin = user_pin

    user_balance = int(input("enter your balance"))
    self.balance = user_balance

    print("pin create successfully")
    self.menu()

  def change_pin(self):
    old_pin = input("enter old pin")

    if old_pin == self.pin:
      new_pin = input("enter new pin")
      self.pin = new_pin
      print("pin change successfully")
      self.menu()

    else:
      print("enter correct old pin")
      self.menu()

  def check_balance(self):
    user_pin = input("enter pin ")
    if user_pin == self.pin:
      print("your balance is" , self.balance)
    else:
      print("you enter wrong pin try agin")
    self.menu()

  def withdraw(self):
    user_pin = input("enter pin")
    if user_pin == self.pin:
      amount = int(input("enter your amount"))
      if amount <= self.balance:
        self.balance = self.balance - amount
        print("your balance is " ,self.balance)
      else:
        print("low balance in your account")
    else:
      print("wrong pin")
    self.menu()





```
